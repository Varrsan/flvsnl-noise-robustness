```python
import os
import json
import shutil

# 1. Clear working directory
for f in os.listdir("/kaggle/working"):
    if f.endswith(".py"):
        os.remove(f"/kaggle/working/{f}")
os.makedirs("/kaggle/working/results_v2", exist_ok=True)

# 2. Copy NEW Stage 2 Python scripts
target_files = ["flvsnl_core.py", "flvsnl_sanity.py", "flvsnl_stage2.py", "flvsnl_stage2b.py", "flvsnl_analyze2.py", "flvsnl_analyze1b.py"]
for root, dirs, files in os.walk("/kaggle/input"):
    for file in files:
        if file in target_files:
            shutil.copy(os.path.join(root, file), os.path.join("/kaggle/working", file))

# 3. Merge ALL json caches we can find to get our 200 runs back!
master_cache = {}
for root, dirs, files in os.walk("/kaggle/input"):
    for file in files:
        if "json" in file:
            with open(os.path.join(root, file), "r") as f:
                data = json.load(f)
                master_cache.update(data)

# 4. Save the master cache
with open("/kaggle/working/results_v2/runs.json", "w") as f:
    json.dump(master_cache, f)

print(f"✅ Scripts copied!")
print(f"✅ Master cache created with {len(master_cache)} unique runs (Should be 200).")
```
